# 06 — Every number, table and figure of the paper

This notebook writes what the LaTeX source in `paper/` reads, so that no number in
the PDF is typed by hand:

- `paper/numbers.tex`: one `\newcommand` per number quoted in the text
  (for example `\nXgbAuroc`), used by `paper/main.tex` and `paper/appendix.tex`;
- `paper/tables/*.tex`: every table;
- `figuras/timeline.pdf`, `figuras/realtime_by_horizon.pdf`,
  `figuras/alert_tradeoff.pdf` and `figuras/calibration_isotonic.pdf` (the other
  figures come from `04_results`);
- `paper/abstract_openreview.txt`: the abstract as plain text.

The code is `src/paper/paper_outputs.py`. Besides writing, it checks every
qualitative claim the text makes about these numbers (for example, that XGBoost
outperforms climatology on onsets at every horizon with intervals that do not
overlap) and stops if one no longer holds. It also checks that every macro the
paper uses is defined.

**Inputs.** `data/train/tables/` (from `04`), `data/train/revision/` (from `05`)
and `data/master/`. Run time: under a minute.

**Building the PDF.** From `paper/`: `pdflatex main && bibtex main && pdflatex main
&& pdflatex main`.

```python
import runpy
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
runpy.run_path(str(PROJECT_ROOT / "src" / "paper" / "paper_outputs.py"), run_name="__main__");
```
